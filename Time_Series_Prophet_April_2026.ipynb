```python
# Code snippet 2
# Importing Prophet package
from prophet import Prophet
```

```python
# Code snippet 3
# Importing the time series data
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/nvamsimohan/DallasDSA/refs/heads/main/Statefarm%20Incidents.csv')

# Display the data
data
```

```python
# Code snippet 4
# Creating 2 additional columns - ds and y (to satisfy the Prophet package rqmt)
data['ds'] = data['Date']
data['y'] = data['Incidents']

data
```

```python
# Code snippet 5
# Starting the Prophet  instance
model = Prophet()
```

```python
# Code snippet 6
# Building time series model using Prophet
model.fit(data[['ds', 'y']])
```

```python
# Code snippet 7
# Predicting the number of Incidents for the next 30 days
future = model.make_future_dataframe(periods=30)
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']]
```
